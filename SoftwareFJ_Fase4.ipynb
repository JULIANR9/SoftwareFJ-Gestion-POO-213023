Este notebook presenta la solución de la Fase 4. El sistema funciona con objetos y listas en memoria, sin utilizar bases de datos. Los eventos y errores se registran en `software_fj.log`.

## 1. Objetivo
Construir una aplicación orientada a objetos para registrar clientes, administrar servicios y gestionar reservas, aplicando validaciones y manejo robusto de excepciones.

```python
# Importamos las clases principales del proyecto.
from entidades import Cliente
from servicios import ReservaSala, AlquilerEquipo, AsesoriaEspecializada
from sistema import SistemaFJ
from excepciones import ErrorSistema
```

## 2. Creación del sistema y registro de clientes
La clase Cliente valida nombre, correo y teléfono. Los datos se encapsulan mediante propiedades.

```python
sistema = SistemaFJ()

# Registro válido.
cliente1 = Cliente('C001', 'Laura Pérez', 'laura@email.com', '3001234567')
sistema.registrar_cliente(cliente1)
print(cliente1.describir())

# Registro inválido: el correo no cumple el formato esperado.
try:
    Cliente('C002', 'Carlos Ruiz', 'correo-invalido', '3111234567')
except ErrorSistema as error:
    print('Error controlado:', error)
```

## 3. Registro de servicios y polimorfismo
Los tres servicios heredan de la clase abstracta Servicio y sobrescriben la validación y el cálculo de costos. El método `calcular_costo` acepta parámetros opcionales para combinar cantidad, impuesto y descuento.

```python
sala = ReservaSala('S001', 'Sala de reuniones', 50000)
equipo = AlquilerEquipo('S002', 'Alquiler de portátil', 30000)
asesoria = AsesoriaEspecializada('S003', 'Asesoría de software', 80000)

for servicio in (sala, equipo, asesoria):
    sistema.registrar_servicio(servicio)
    print(servicio.describir(), '| costo base:', servicio.calcular_costo())
```

## 4. Creación y procesamiento de reservas
Una reserva pasa por los estados CREADA, CONFIRMADA y PROCESADA. También puede cancelarse antes de ser procesada.

```python
reserva = sistema.crear_reserva('R001', 'C001', 'S001', 2)
reserva.confirmar()
total = sistema.procesar_reserva('R001', impuesto=0.19, descuento=0.10)
print(reserva.describir())
print(f'Total calculado: ${total:,.2f}')
```

## 5. Simulaciones de errores y continuidad del sistema
Se ejecutan casos válidos e inválidos. Los errores se capturan para que la aplicación continúe y se registran en el archivo de logs.

```python
casos = [
    ('Correo inválido', lambda: Cliente('C010', 'Ana Pérez', 'ana-sin-arroba', '3001112233')),
    ('Servicio inexistente', lambda: sistema.crear_reserva('R010', 'C001', 'S999', 1)),
    ('Cantidad inválida', lambda: sistema.crear_reserva('R011', 'C001', 'S002', 0)),
    ('Procesamiento sin confirmar', lambda: sistema.crear_reserva('R012', 'C001', 'S003', 1).procesar()),
    ('Cancelar reserva procesada', lambda: sistema.cancelar_reserva('R001')),
]
for nombre, operacion in casos:
    try:
        operacion()
        print(nombre, ': operación ejecutada')
    except Exception as error:
        print(nombre, ': error controlado ->', error)

print('El sistema continúa activo.')
```

## 6. Resumen y archivo de logs
El sistema utiliza el módulo `logging` para registrar eventos y errores en `software_fj.log`. La clase gestora mantiene las listas de clientes, servicios y reservas en memoria.

```python
print('Resumen:', sistema.resumen())
print('Consulte el archivo software_fj.log para revisar los registros.')
```
